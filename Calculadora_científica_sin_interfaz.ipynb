```python
import math

def sumar(a,b):
    return (a+b)

def restar(a,b):
    return(a-b)

def multiplicar(a,b):
    return(a*b)

def dividir(a,b):
    if b == 0: ## Por si el usuario mete un cero para dividir
        return "No se puede dividir entre cero"
    return(a/b)

def potencia(a,b):
    return(a**b)

def raiz(a):
    if a < 0: ## No puede haber raices de números negativos
        return "No se puede hacer la raiz de un número negativo"
    return math.sqrt(a)

def seno_grados(a):
    radianes = math.radians(a)
    return math.sin(radianes)

def logaritmo(a):
    if a <= 0:
        return "El logaritmo debe ser con números mayores que cero"
    return math.log10(a)
```
