```python
import math
import tkinter as tk
from tkinter import messagebox

## FUNCIONES

def sumar(a,b):
    return (a+b)

def restar(a,b):
    return(a-b)

def multiplicar(a,b):
    return(a*b)

def dividir(a,b):
    if b == 0: ## Por si el usuario mete un cero para dividir
        return "No se puede dividir entre cero"
    return(a/b)

def potencia(a,b):
    return(a**b)

def raiz(a):
    if a < 0: ## No puede haber raices de números negativos
        return "No se puede hacer la raiz de un número negativo"
    return math.sqrt(a)

def seno_grados(a):
    radianes = math.radians(a)
    return math.sin(radianes)

def logaritmo(a):
    if a <= 0:
        return "El logaritmo debe ser con números mayores que cero"
    return math.log10(a)

## LÓGICA INTERFAZ CALCULADORA

def obtener_valores():
    """Intenta leer los números de las cajas de texto."""
    try:
        a = float(entry_a.get()) if entry_a.get() else 0.0
        b = float(entry_b.get()) if entry_b.get() else 0.0
        return a, b
    except ValueError:
        messagebox.showerror("Error", "Por favor, introduce números válidos.")
        return None

def ejecutar_operacion(tipo):
    """Llama a la función matemática y muestra el resultado."""
    valores = obtener_valores()
    if valores is None:
        return

    a, b = valores

    if tipo == "sumar":
        res = sumar(a, b)
    elif tipo == "restar":
        res = restar(a, b)
    elif tipo == "multiplicar":
        res = multiplicar(a, b)
    elif tipo == "dividir":
        res = dividir(a, b)
    elif tipo == "potencia":
        res = potencia(a, b)
    elif tipo == "raiz":
        res = raiz(a)  # Solo usa 'a'
    elif tipo == "seno":
        res = seno_grados(a)  # Solo usa 'a'
    elif tipo == "logaritmo":
        res = logaritmo(a)  # Solo usa 'a'

    label_resultado.config(text=f"Resultado: {res}")

def limpiar():
    """Borra el contenido de las entradas y reinicia el resultado."""
    entry_a.delete(0, tk.END)
    entry_b.delete(0, tk.END)
    label_resultado.config(text="Resultado: ")

## DISEÑO VENTANA INTERFAZ

ventana = tk.Tk()
ventana.title("Calculadora Científica con interfaz Tkinter")
ventana.geometry("350x400")

# Entradas de texto para los números
tk.Label(ventana, text="Número A (o Ángulo):").pack(pady=5)
entry_a = tk.Entry(ventana)
entry_a.pack()

tk.Label(ventana, text="Número B (si aplica):").pack(pady=5)
entry_b = tk.Entry(ventana)
entry_b.pack()

# Contenedor para organizar los botones en cuadrícula
frame_botones = tk.Frame(ventana)
frame_botones.pack(pady=20)

# Botones de las operaciones
tk.Button(frame_botones, text="+", width=10, command=lambda: ejecutar_operacion("sumar")).grid(row=0, column=0, padx=5, pady=5)
tk.Button(frame_botones, text="-", width=10, command=lambda: ejecutar_operacion("restar")).grid(row=0, column=1, padx=5, pady=5)
tk.Button(frame_botones, text="*", width=10, command=lambda: ejecutar_operacion("multiplicar")).grid(row=1, column=0, padx=5, pady=5)
tk.Button(frame_botones, text="/", width=10, command=lambda: ejecutar_operacion("dividir")).grid(row=1, column=1, padx=5, pady=5)
tk.Button(frame_botones, text="Potencia", width=10, command=lambda: ejecutar_operacion("potencia")).grid(row=2, column=0, padx=5, pady=5)
tk.Button(frame_botones, text="Raíz A", width=10, command=lambda: ejecutar_operacion("raiz")).grid(row=2, column=1, padx=5, pady=5)
tk.Button(frame_botones, text="Seno A", width=10, command=lambda: ejecutar_operacion("seno")).grid(row=3, column=0, padx=5, pady=5)
tk.Button(frame_botones, text="Log10 A", width=10, command=lambda: ejecutar_operacion("logaritmo")).grid(row=3, column=1, padx=5, pady=5)

# Etiqueta para mostrar el resultado final
label_resultado = tk.Label(ventana, text="Resultado: ", font=("Arial", 12, "bold"))
label_resultado.pack(pady=20)

# Iniciar la aplicación
ventana.mainloop()
```
