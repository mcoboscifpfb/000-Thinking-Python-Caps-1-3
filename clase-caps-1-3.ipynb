## Parte 3 — La tripulación trabaja (capítulo 3)

Llevamos toda la clase repitiendo cálculos. Cuando algo se repite tres veces, es que hace falta una **función**: empaquetar sentencias bajo un nombre y usarlo cuando haga falta.

Definir una función no la ejecuta. Solo le enseñamos a Python un verbo nuevo.

```python
def despegar():
    print("Cuenta atrás...")
    print("3... 2... 1...")
    print("Ignición")
```

```python
# esto sí la ejecuta: llamarla, con sus paréntesis
despegar()
```

### Parámetros y argumentos

Si la función necesita un dato de entrada, declara un **parámetro**. El valor concreto que le pasamos al llamarla es el **argumento**.

```python
def transmitir(mensaje):
    print("Houston dice:", mensaje)

transmitir("todo nominal")
```

```python
# el argumento puede ser cualquier expresión; se evalúa antes de entrar
transmitir("quedan " + str(60) + " minutos de ventana")
```

### Repetir cosas: el bucle for dentro de la función

Si metemos un `for` dentro del `def`, el cuerpo entero se repite. Aquí es donde las funciones empiezan a compensar de verdad.

```python
def comprobar_sistemas():
    print("  motor ok")
    print("  oxígeno ok")
    print("  comunicaciones ok")

for i in range(3):
    print(f"Comprobación {i + 1}")
    comprobar_sistemas()
```

### Variables locales y la pila

Lo que se crea dentro de una función se queda dentro. Los parámetros y las variables de la función son **locales**: cuando la función termina, desaparecen.

```python
def autonomia(litros, consumo):
    horas = litros / consumo
    print(f"Autonomía: {horas} horas")

autonomia(84, 5)
```

```python
# horas no existe aquí fuera
horas
```

Cada vez que se llama a una función, Python apila un *frame* nuevo con sus variables encima del principal, y lo retira al acabar. Eso es el **diagrama de pila** del libro, y es también la razón de que los tracebacks tengan esa forma de escalera:

```python
def nivel_oxigeno():
    return 100 / 0     # el ingeniero escribió mal la fórmula

nivel_oxigeno()
```

Mirad el rastro: primero la línea desde la que se llamó, luego la línea exacta del fallo dentro de la función. Leer tracebacks de abajo arriba ahorra muchas horas.

### Thinkful - Number Drills: Quadratic formula (7 kyu)

Dados `a`, `b` y `c` de una ecuación de segundo grado, hay que devolver las dos raíces en una lista, de menor a mayor. Aquí juntamos varias cosas de hoy: parámetros, `math.sqrt`, y la fórmula de toda la vida, `x = (-b ± √(b² − 4ac)) / 2a`.

```python
import math

def quadratic_formula(a, b, c):
    discriminante = math.sqrt(b ** 2 - 4 * a * c)
    x1 = (-b + discriminante) / (2 * a)
    x2 = (-b - discriminante) / (2 * a)
    return [min(x1, x2), max(x1, x2)]

quadratic_formula(1, -3, 2)   # x² - 3x + 2 = 0  ->  [1.0, 2.0]
```

### Sentence Smash (8 kyu)

Nos dan una lista de palabras y hay que unirlas en una frase, con un espacio entre cada una y sin espacios extra al principio ni al final. El método `join` de los strings hace exactamente eso: el string sobre el que se llama (`" "`) es el pegamento.

```python
def smash(words):
    return " ".join(words)

smash(["hola", "clase", "esto", "es", "Python"])
```
