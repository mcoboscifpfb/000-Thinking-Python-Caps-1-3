```python
import math
import tkinter as tk
from tkinter import messagebox

# ==========================================
# TUS FUNCIONES MATEMÁTICAS
# ==========================================
def sumar(a, b): return a + b
def restar(a, b): return a - b
def multiplicar(a, b): return a * b
def dividir(a, b): return "No se puede dividir entre cero" if b == 0 else a / b
def potencia(a, b): return a ** b
def raiz(a): return "No se puede hacer la raíz de un número negativo" if a < 0 else math.sqrt(a)
def seno_grados(a): return math.sin(math.radians(a))
def logaritmo(a): return "El logaritmo debe ser con números mayores que cero" if a <= 0 else math.log10(a)

# ==========================================
# LÓGICA DE SECUENCIA (A -> OPERADOR -> B)
# ==========================================
numero_a = None
operador = None

def click_numero(valor):
    """Añade el número presionado a la pantalla."""
    pantalla.insert(tk.END, valor)

def seleccionar_operador(op):
    """Paso 2: Guarda el Número A y el Operador seleccionado o gestiona el signo menos."""
    global numero_a, operador
    try:
        contenido = pantalla.get()

        # SI LA PANTALLA ESTÁ VACÍA Y SE PULSA "-", lo usamos como signo negativo del número
        if contenido == "" and op == "-":
            pantalla.insert(tk.END, "-")
            return

        # Si ya hay un número escrito, actúa normalmente guardando el operador
        if contenido != "" and contenido != "-":
            numero_a = float(contenido)
            operador = op
            pantalla.delete(0, tk.END)  # Limpia la pantalla para recibir el Número B
    except ValueError:
        messagebox.showerror("Error", "Introduce un número A válido primero.")

def calcular_resultado():
    """Paso 3: Toma el Número B y ejecuta la operación final."""
    global numero_a, operador
    try:
        if numero_a is None or operador is None:
            return

        # Permitimos procesar números negativos también en la variable B
        contenido_b = pantalla.get()
        if contenido_b == "" or contenido_b == "-":
            numero_b = 0.0
        else:
            numero_b = float(contenido_b)

        # Ejecución según el operador guardado
        if operador == "+": res = sumar(numero_a, numero_b)
        elif operador == "-": res = restar(numero_a, numero_b)
        elif operador == "*": res = multiplicar(numero_a, numero_b)
        elif operador == "/": res = dividir(numero_a, numero_b)
        elif operador == "^": res = potencia(numero_a, numero_b)

        pantalla.delete(0, tk.END)
        pantalla.insert(0, str(res))

        # Reiniciamos las variables de control
        numero_a, operador = None, None
    except ValueError:
        messagebox.showerror("Error", "Introduce un número B válido.")

def calcular_directo(tipo):
    """Para funciones que NO necesitan número B (Raíz, Seno, Log)."""
    try:
        val = float(pantalla.get())
        if tipo == "raiz": res = raiz(val)
        elif tipo == "seno": res = seno_grados(val)
        elif tipo == "log": res = logaritmo(val)

        pantalla.delete(0, tk.END)
        pantalla.insert(0, str(res))
    except ValueError:
        messagebox.showerror("Error", "Introduce un número válido en la pantalla.")

def limpiar():
    global numero_a, operador
    numero_a, operador = None, None
    pantalla.delete(0, tk.END)

# ==========================================
# DISEÑO DE LA CALCULADORA TRADICIONAL
# ==========================================
ventana = tk.Tk()
ventana.title("Calculadora Científica Secuencial")
ventana.geometry("300x420")
ventana.configure(bg="#f4f4f4")

pantalla = tk.Entry(ventana, font=("Arial", 18), justify="right", bd=5, relief=tk.RIDGE)
pantalla.pack(padx=10, pady=15, fill=tk.X)

frame_botones = tk.Frame(ventana, bg="#f4f4f4")
frame_botones.pack()

tk.Button(frame_botones, text="Raíz", width=6, height=2, font=("Arial", 10), command=lambda: calcular_directo("raiz")).grid(row=0, column=0, padx=3, pady=3)
tk.Button(frame_botones, text="Seno", width=6, height=2, font=("Arial", 10), command=lambda: calcular_directo("seno")).grid(row=0, column=1, padx=3, pady=3)
tk.Button(frame_botones, text="Log10", width=6, height=2, font=("Arial", 10), command=lambda: calcular_directo("log")).grid(row=0, column=2, padx=3, pady=3)
tk.Button(frame_botones, text="C", width=6, height=2, font=("Arial", 10, "bold"), bg="#f0a8a8", command=limpiar).grid(row=0, column=3, padx=3, pady=3)

botones_grid = [
    ('7', 1, 0), ('8', 1, 1), ('9', 1, 2), ('/', 1, 3),
    ('4', 2, 0), ('5', 2, 1), ('6', 2, 2), ('*', 2, 3),
    ('1', 3, 0), ('2', 3, 1), ('3', 3, 2), ('-', 3, 3),
    ('0', 4, 0), ('.', 4, 1), ('^', 4, 2), ('+', 4, 3)
]

for (texto, fila, col) in botones_grid:
    if texto in ['/', '*', '-', '+', '^']:
        cmd = lambda t=texto: seleccionar_operador(t)
        color = "#e0e0e0"
    else:
        cmd = lambda t=texto: click_numero(t)
        color = "#ffffff"

    tk.Button(frame_botones, text=texto, width=6, height=2, font=("Arial", 12), bg=color, command=cmd).grid(row=fila, column=col, padx=3, pady=3)

tk.Button(ventana, text="=", width=26, height=2, font=("Arial", 12, "bold"), bg="#a8dbf0", command=calcular_resultado).pack(pady=10)

ventana.mainloop()
```
